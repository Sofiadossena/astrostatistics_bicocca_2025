```python
import numpy as np # type: ignore # type: ignore
import matplotlib.pyplot as plt # type: ignore
import random
from random import sample

k=1000 #numero di prove

p=[1,28,48,98,498] #possibili porte aperte dal presentatore
s=[]
c=[]
n=[]

for j in p:
    switcher=0
    conservative=0
    newcomer=0
    for i in range(k):
    
        #CON DELETE ELIMINA L'INDICE E NON IL VALORE
        N=j+2 #numero di porte 
        x = np.arange(N)

        pos=np.random.randint(0,N) #posizione del premio  

        choise1=np.random.randint(0,N) #prima scelta e scelta conservativa

        el=np.array([pos,choise1])
        y=np.delete(x,el) #porte che può scegliere il presentatore
        choicep=random.sample(list(y),j) #porte aperte dal presentatore
        
        el2=np.append(choise1,choicep) #il secondo non sceglie le porte del presentatore e la sua 
        z=np.delete(x,el2) 
        choice2=np.random.choice(z) #scelta del concorrente che cambia porta

        w=np.delete(x,choicep) 
        choice3=np.random.choice(w) 

        if pos==choise1:
            conservative+=1
        if pos==choice2:
            switcher+=1
        if pos==choice3:
            newcomer+=1


    c.append(conservative/k)
    s.append(switcher/k)
    n.append(newcomer/k)    

plt.plot(p,c,label='Conservative' )
plt.plot(p,s,label='Switcher')
plt.plot(p,n,label='Newcomer') 
plt.legend(loc='upper right')
plt.show()

```
