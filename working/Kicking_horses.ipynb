```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
import random
from random import sample
import math



n = [0, 1, 2, 3, 4]
counts = [109, 65, 22, 3, 1]

mu = np.sum(np.multiply(n,counts))/np.sum(counts)
sigma = np.sqrt(mu)

norm_counts = counts/np.sum(counts)

def poisson(k,mu):
    return mu**k*np.exp(-mu)/math.factorial(k)

plt.bar(n, norm_counts, color='b', label='Data')
plt.scatter(n, norm_counts, color='r', label='Data')
plt.plot(n, [poisson(k,mu) for k in n], label='Poisson')

```
