```python
import numpy as np  
import matplotlib.pyplot as plt 
import random
from random import sample
import scipy
import astropy.visualization.hist

today=0 #cloudy day 
cloudy_counts=0
sunny_counts=0
Ps=[]
Pc=[]

N=100000

for i in range(1,N):
    if today ==0:
        p=np.random.uniform(0,1)
        if p<=0.5:
            today=0
        else :
            today=1
    else :
        p=np.random.uniform(0,1)
        if p<=0.1:
            today=0
        else :
            today=1
    if today ==0:
        cloudy_counts+=1
    else :
        sunny_counts+=1
    
    Ps.append(sunny_counts/i)
    Pc.append(cloudy_counts/i)

Psunny= sunny_counts/N
Pcloudy= cloudy_counts/N

x=np.linspace(1,N,N-1)
plt.plot(x,Ps, label= "Probability of sunny day")
plt.plot(x,0.83*np.ones(len(x)), '--', color='black')
plt.xlabel("Days")
plt.ylabel("Probability")
plt.legend()
plt.show()
plt.plot(x,Pc, label= "Probability of cloudy day")
plt.plot(x,0.17*np.ones(len(x)), '--', color='black')
plt.xlabel("Days")
plt.ylabel("Probability")
plt.legend()
plt.show()
plt.plot(x[40000:],Ps[40000:], label= "Probability of sunny day")
plt.xlabel("Days")
plt.ylabel("Probability")
plt.legend()
plt.show()
plt.plot(x[40000:],Pc[40000:], label= "Probability of cloudy day")
plt.xlabel("Days")
plt.ylabel("Probability")
plt.legend()
plt.show()
```

### Changing the burn-in phase 
Hist of sunny days with all the steps 

```python
MeanPs= np.mean(Ps)
DevPs= np.std(Ps)
plt.hist(Ps, 100, density=True, histtype='step', label='mean = %.4f, std = %.4f' % (MeanPs, DevPs))
astropy.visualization.hist(Ps, bins="freedman", histtype="step",density=True, label='freedman') #using freedman bins
#plt.xlim(0.82,0.85) 
plt.legend(loc="best")
plt.show()

```
